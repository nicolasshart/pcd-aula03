dados

```python
import numpy as np

dados = np.array([
    [120, 15000, 100, 125.00, 4],
    [135, 16200, 108, 120.00, 5],
    [98,  11700,  82, 119.39, 3],
    [150, 21000, 125, 140.00, 7],
    [175, 24500, 140, 140.00, 9],
    [160, 22400, 132, 140.00, 6],
    [110, 13200,  95, 120.00, 2]
])
dias = np.array(["Segunda","Terça","Quarta","Quinta","Sexta","Sábado","Domingo"])
nomes = np.array(["Pedidos","Faturamento","Clientes","Ticket","Reclamações"])

# colunas
ped, fat, cli, tick, rec = dados[:,0], dados[:,1], dados[:,2], dados[:,3], dados[:,4]
```

parte 1

```python
print("linhas:", dados.shape[0], "| colunas:", dados.shape[1])
print("total:", dados.size, "| dimensões:", dados.ndim, "| dtype:", dados.dtype)
np.set_printoptions(suppress=True)
```

parte 2

```python
print("2.1 sexta:", dados[4])
print("2.2 faturamento:", dados[:,1])
print("2.3 reclamações:", dados[:,4])
print("2.4 quinta a domingo:\n", dados[3:])
print("2.5 pedidos, fat., clientes:\n", dados[:, :3])
```

parte 3

```python
print("3.1 total pedidos:", np.sum(ped))
print("3.2 faturamento total:", np.sum(fat))
print("3.3 total clientes:", np.sum(cli))
print("3.4 média pedidos:", np.mean(ped))
print("3.5 média faturamento:", np.mean(fat))
print("3.6 média reclamações:", np.mean(rec))
```

parte 4

```python
print("4.1 mais pedidos:", dias[np.argmax(ped)], "| menos:", dias[np.argmin(ped)])
print("4.2 maior fat.:", dias[np.argmax(fat)], "| menor:", dias[np.argmin(fat)])
print("4.3 amplitude:", np.max(fat) - np.min(fat))
```

parte 5

```python
print("5.1 fat. > 20000:", dias[fat > 20000])
print("5.2 pedidos > 140:", dias[ped > 140])
print("5.3 reclamações > 6:", dias[rec > 6])
print("5.4 fat. com pedidos > 150:", fat[ped > 150])
print("5.5 abaixo da média:", np.sum(fat < np.mean(fat)))  # true conta 1
```

parte 6

```python
classificacao = np.where(fat >= 20000, "Alto desempenho", "Desempenho normal")
print("6.1", np.column_stack((dias, classificacao)))
print("6.2 alto desempenho:", np.sum(classificacao == "Alto desempenho"))
```

parte 7

```python
media, desvio = np.mean(fat), np.std(fat)
print("média:", media, "| mediana:", np.median(fat), "| desvio:", desvio)
print("mín:", np.min(fat), "| máx:", np.max(fat))
print("coef. variação:", desvio / media)
```

Tem grande variação. Média de 17.714, mediana de 16.200 e desvio de 4.551. O maior faturamento de 24.500 é mais que o dobro do menor de 11.700.

parte 8

```python
print("8.1 médias:", np.mean(dados, axis=0))
print("8.2 máximos:", np.max(dados, axis=0))
print("8.3 mínimos:", np.min(dados, axis=0))

# variação relativa = desvio / média
cv = np.std(dados, axis=0) / np.mean(dados, axis=0)
print("8.4 cv:", cv)
print("maior variação:", nomes[np.argmax(cv)])
```

Tem a maior variação de 0,43, seguida pelo faturamento de 0,26. O ticket médio é o mais estável de 0,07.

parte 9

```python
i = np.argmax(fat)
print("9.1 melhor dia:", dias[i])
print("9.2 também mais pedidos?", i == np.argmax(ped))
print("9.3 também mais reclamações?", i == np.argmax(rec))
print("9.4 >150 pedidos e fat. >20000:", np.sum((ped > 150) & (fat > 20000)))
```

parte 10

```python
print("RESUMO DA SEMANA")
print(f"total de pedidos: {np.sum(ped):.0f}")
print(f"faturamento total: R$ {np.sum(fat):,.2f}")
print(f"média diária de pedidos: {np.mean(ped):.2f}")
print(f"média diária de faturamento: R$ {np.mean(fat):,.2f}")
print(f"melhor dia: {dias[np.argmax(fat)]}")
print(f"pior dia: {dias[np.argmin(fat)]}")
print(f"maior nº de reclamações: {dias[np.argmax(rec)]} ({rec.max():.0f})")
print(f"dias de alto desempenho: {np.sum(fat >= 20000)}")
```

parte 11 apoio

```python
# correlação com reclamações
print("pedidos x reclamações:", np.corrcoef(ped, rec)[0,1])
print("pedidos x faturamento:", np.corrcoef(ped, fat)[0,1])
```

Teve desempenho diferent com faturamento total de 124.000, concentrado de quinta a sábado com todos acima de 20.000. Os outros quatro dias abaixo da média. Na quarta, pior dia com 98 pedidos e 11.700, e na sexta, melhor em faturamento de 24.500, mas também com mais 9 reclamações.
Pedidos e faturamento têm relação de 0,99 e pedidos e reclamações, 0,94. Quanto mais pedidos, mais reclamações. Com só 7 dias, isso é indício.
