```python
import google_colab_selenium as gs
from selenium.webdriver.common.by import By
import json
```

```python
cidade = input("Digite o nome da cidade: ")
```

```python
navegador = gs.Chrome()
```

```python
url = f"https://wttr.in/{cidade}?format=j1"
```

```python
navegador.get(url)
```

```python
texto = navegador.find_element(By.TAG_NAME, "body").text
```

```python
dados = json.loads(texto)
previsao_hoje = dados["weather"][0]
```

```python
maxima = previsao_hoje["maxtempC"]
minima = previsao_hoje["mintempC"]
```

```python
print(f"Previsão para {cidade}")
print("Temperatura máxima:", maxima, "°C")
print("Temperatura mínima:", minima, "°C")
```

```python
navegador.quit()
```
