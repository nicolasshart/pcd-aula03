```python
import pandas as pd

df = pd.read_csv('https://raw.githubusercontent.com/andr3nf/clima_municipios_brasileiros/refs/heads/main/clima_historico_municipal.csv')

cidade = df[(df['nome_municipio'] == 'Santa Rosa') & (df['uf'] == 'RS')]

# todas as informações
cidade

# mês de temperatura mínima
meses = ['jan','fev','mar','abr','mai','jun','jul','ago','set','out','nov','dez']
temp_cols = [f'temp_media_{m}' for m in meses]
mes_temp_min = cidade[temp_cols].T.idxmin().values[0]
print(mes_temp_min)

# mês de maior déficit hídrico
def_cols = [f'def_{m}' for m in meses]
mes_def_max = cidade[def_cols].T.idxmax().values[0]
print(mes_def_max)

# média de temperatura pelas 4 estações
verao = ['temp_media_dez','temp_media_jan','temp_media_fev']
outono = ['temp_media_mar','temp_media_abr','temp_media_mai']
inverno = ['temp_media_jun','temp_media_jul','temp_media_ago']
primavera = ['temp_media_set','temp_media_out','temp_media_nov']

print('verao:', cidade[verao].mean(axis=1).values[0])
print('outono:', cidade[outono].mean(axis=1).values[0])
print('inverno:', cidade[inverno].mean(axis=1).values[0])
print('primavera:', cidade[primavera].mean(axis=1).values[0])
```
