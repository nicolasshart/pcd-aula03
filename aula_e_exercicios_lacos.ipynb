Exercício 1

```python
numero = int(input("digite um número: "))
primo = True

if numero < 2:
    primo = False
elif numero == 2:
    primo = True
elif numero % 2 == 0:
    primo = False
else:
    i = 3
    while i < numero and primo:
        if numero % i == 0:
            primo = False
        i += 2

if primo:
    print("{} é primo".format(numero))
else:
    print("{} não é primo".format(numero))
```

Exercício 2

```python
def eh_primo(numero):
    if numero < 2:
        return False
    if numero == 2:
        return True
    if numero % 2 == 0:
        return False
    i = 3
    while i < numero:
        if numero % i == 0:
            return False
        i += 2
    return True

quantidade = int(input("quantos números primos deseja listar? "))
primos = []
numero = 2

while len(primos) < quantidade:
    if eh_primo(numero):
        primos.append(numero)
    numero += 1

print(primos)
```
