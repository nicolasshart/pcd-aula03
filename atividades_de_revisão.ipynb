atividade i

```python
import random

numero = random.randint(0, 15)
tentativas = 0
historico = []

while True:
    chute = int(input("digite um numero de 0 a 15 (ou -1 para parar): "))
    if chute == -1:
        break
    tentativas += 1
    if chute == numero:
        print("acertou!")
        historico.append((numero, tentativas))
        with open("historico.txt", "a") as f:
            f.write(f"numero: {numero} tentativas: {tentativas}\n")
        numero = random.randint(0, 15)
        tentativas = 0
    elif chute < numero:
        print("o numero e maior")
    else:
        print("o numero e menor")

print("historico da sessao:")
for item in historico:
    print(item)
```

atividade ii

```python
import csv

lista = []
total = 0

while True:
    item = input("nome do item (ou 'sair' para terminar): ")
    if item == "sair":
        break
    quantidade = int(input("quantidade: "))
    valor = float(input("valor unitario: "))
    subtotal = quantidade * valor
    total += subtotal
    lista.append([item, quantidade, valor, subtotal])

print(f"valor total das compras: {total:.2f}")

with open("compras.csv", "w", newline="") as arquivo:
    escritor = csv.writer(arquivo)
    escritor.writerow(["item", "quantidade", "valor", "subtotal"])
    for linha in lista:
        escritor.writerow(linha)
```
